## Zonal statistics of seasonal vegetation composite by forest type

## Data source: Landsat 5

### Vegetation Index: NDVI

```python
composite_strategy = "Max" # Max, Mean, Median
vegetation_index = "NDVI"; # NDVI, NDMI, EVI

forest_type_labels = {1: "Cool", 2: "Intermediate" , 3: "Hot"}
forest_type_raster = "fortype_aoiN_UTM_30m"

data_dry, labels = utils.zonal_stats(f"{composite_strategy}_{vegetation_index}_2002_06To08_North_30m", forest_type_raster, forest_type_labels)
data_wet, labels = utils.zonal_stats(f"{composite_strategy}_{vegetation_index}_2003_06To08_North_30m", forest_type_raster, forest_type_labels)
legend_labels = ["2002 (Dry)", "2003 (Wet)"]
```

```python
fig, ax = utils.create_boxplot(data_dry, data_wet, labels, legend_labels, composite_strategy, vegetation_index);
```

```python
utils.create_histogram(data_dry, data_wet, labels, legend_labels, composite_strategy, vegetation_index);
```

```python
utils.create_gaussian_kde(data_dry, data_wet, labels, legend_labels, composite_strategy, vegetation_index);
```

```python
fig, ax = utils.create_bar_chart(data_dry, data_wet, labels, legend_labels, composite_strategy, vegetation_index);
ax.set_ylim(0.8,1);
```
